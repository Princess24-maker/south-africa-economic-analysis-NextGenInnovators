## Step 7: Reshape the Poverty Dataset
We need long format so that the year becomes a row(this will make merging and analysis easier)

```python
poverty_year_columns = [
    col for col in poverty_sa.columns
    if str(col).isdigit() and 1900 <= int(col) <= 2025
]

poverty_long = poverty_sa.melt(
    id_vars=[
        "REF_AREA",
        "REF_AREA_LABEL",
        "INDICATOR",
        "INDICATOR_LABEL"
    ],
    value_vars=poverty_year_columns,
    var_name="Year",
    value_name="Poverty"
)

poverty_long["Year"] = pd.to_numeric(
    poverty_long["Year"],
    errors="coerce"
)

poverty_long["Poverty"] = pd.to_numeric(
    poverty_long["Poverty"],
    errors="coerce"
)

print("Poverty data reshaped successfully.")
display(poverty_long.head())
```

### 7.2 Checking Missing Poverty Observations

```python
poverty_missing = poverty_long["Poverty"].isna().sum()
poverty_total = len(poverty_long)

print("Total poverty observations:", poverty_total)
print("Missing poverty observations:", poverty_missing)
print(
    "Missing percentage:",
    round((poverty_missing / poverty_total) * 100, 2),
    "%"
)
```

### 7.3 Handling Missing Poverty Observations

```python
poverty_clean = poverty_long.dropna(
    subset=["Year", "Poverty"]
).copy()

print("Observations before cleaning:", len(poverty_long))
print("Observations after cleaning:", len(poverty_clean))
print(
    "Observations removed:",
    len(poverty_long) - len(poverty_clean)
)
```

### 7.3.1 Justification for Cleaning Decisions

**Why we excluded missing poverty years:**
- Poverty statistics are collected via occasional household surveys, not annually.
- There are only 7 valid observations from 1993 to 2022
- Each survey provides is an independent measurement
- Filling in missing years would introduce artificiality into the data
- It is better to analyse the actual observed data

**Why we dropped missing inflation years:**
- Only 2 years are missing (1981, 1982)
- The gaps are most likely due to limitations in early
- Minimal effect on overall analysis
- No need for imputation

**Why we used an inner join for merging:**
- This only includes years present in both datasets
- Guarantees a valid comparison between poverty and inflation measures
- Result: 7 shared years (1993, 2000, 2005, 2008, 2010, 2014, 2022)

### 7.4 Validating the Cleaned Poverty Dataset

```python
print("Missing values:")
print(poverty_clean[["Year", "Poverty"]].isna().sum())

print("\nDuplicate rows:", poverty_clean.duplicated().sum())

print("\nYear range:")
print(
    int(poverty_clean["Year"].min()),
    "to",
    int(poverty_clean["Year"].max())
)

print("\nData types:")
print(poverty_clean[["Year", "Poverty"]].dtypes)
```

## Step 8: Reshaping the Inflation Dataset

### 8.1 Converting Wide Format to Long Format

```python
inflation_year_columns = [
    col for col in inflation_actual.columns
    if str(col).isdigit() and 1900 <= int(col) <= 2025
]

print("Number of inflation year columns:", len(inflation_year_columns))
print("First years:", inflation_year_columns[:5])
print("Last years:", inflation_year_columns[-5:])
```

### 8.2 Converting Inflation Data to Long Format

```python
inflation_long = inflation_actual.melt(
    id_vars=[
        "REF_AREA",
        "REF_AREA_LABEL",
        "INDICATOR",
        "INDICATOR_LABEL"
    ],
    value_vars=inflation_year_columns,
    var_name="Year",
    value_name="Inflation"
)

inflation_long["Year"] = pd.to_numeric(
    inflation_long["Year"],
    errors="coerce"
)

inflation_long["Inflation"] = pd.to_numeric(
    inflation_long["Inflation"],
    errors="coerce"
)

print("Inflation data reshaped successfully.")
display(inflation_long.head())
```

### 8.3 Checking Missing Inflation Observations

```python
inflation_missing = inflation_long["Inflation"].isna().sum()
inflation_total = len(inflation_long)

print("Total inflation observations:", inflation_total)
print("Missing inflation observations:", inflation_missing)

print(
    "Missing percentage:",
    round((inflation_missing / inflation_total) * 100, 2),
    "%"
)
```

### 8.4 Handling Missing Inflation Values

Missing inflation observations are removed only where the actual inflation value is unavailable. The missing values are not replaced with zero because zero inflation represents a valid economic observation, whereas a missing value represents unavailable data.

```python
inflation_clean = inflation_long.dropna(
    subset=["Year", "Inflation"]
).copy()

print("Observations before cleaning:", len(inflation_long))
print("Observations after cleaning:", len(inflation_clean))
print(
    "Observations removed:",
    len(inflation_long) - len(inflation_clean)
)
```

### 8.5 Validating the Cleaned Inflation Dataset

The cleaned inflation dataset is validated to confirm that missing analytical values have been addressed, duplicate records are absent, the year range is valid, and the Year and Inflation variables have appropriate data types.

```python
print("Missing values:")
print(inflation_clean[["Year", "Inflation"]].isna().sum())

print("\nDuplicate rows:", inflation_clean.duplicated().sum())

print("\nYear range:")
print(
    int(inflation_clean["Year"].min()),
    "to",
    int(inflation_clean["Year"].max())
)

print("\nData types:")
print(inflation_clean[["Year", "Inflation"]].dtypes)
```

## Step 9: Preparing the Datasets for Integration

### 9.1 Comparing Year Coverage

```python
print("POVERTY")
print("Minimum year:", int(poverty_clean["Year"].min()))
print("Maximum year:", int(poverty_clean["Year"].max()))
print("Valid observations:", len(poverty_clean))

print("\nINFLATION")
print("Minimum year:", int(inflation_clean["Year"].min()))
print("Maximum year:", int(inflation_clean["Year"].max()))
print("Valid observations:", len(inflation_clean))
```

### 9.2 Identifying Common Years

```python
poverty_year_set = set(poverty_clean["Year"])
inflation_year_set = set(inflation_clean["Year"])

common_years = sorted(
    poverty_year_set.intersection(inflation_year_set)
)

print("Number of common years:", len(common_years))
print("Common year range:", min(common_years), "to", max(common_years))
print("\nCommon years:")
print(common_years)
```

### 9.3 Checking for Unique Yearly Observations

Before combining the datasets, the number of observations per year is checked to ensure that each year has a single valid value. This prevents duplicated records from distorting the merged analysis.

```python
print("POVERTY OBSERVATIONS PER YEAR")
print(poverty_clean["Year"].value_counts().sort_index())

print("\nINFLATION OBSERVATIONS PER YEAR")
print(inflation_clean["Year"].value_counts().sort_index())
```

### 9.4 Aligning the Datasets to Common Years

The poverty and inflation datasets are aligned using their common years. This ensures that each observation in the final dataset represents the same year for both indicators.

```python
poverty_common = poverty_clean[
    poverty_clean["Year"].isin(common_years)
].copy()

inflation_common = inflation_clean[
    inflation_clean["Year"].isin(common_years)
].copy()

print("Poverty observations for common years:", len(poverty_common))
print("Inflation observations for common years:", len(inflation_common))
```

## Step 10: Integrating the South African Datasets

### 10.1 Merging Poverty and Inflation by Year

The two cleaned South African datasets are merged using Year as the common identifier. An inner join is used so that only years with valid observations in both datasets are included in the integrated dataset.

```python
combined_sa = pd.merge(
    poverty_common[
        ["Year", "Poverty"]
    ],
    inflation_common[
        ["Year", "Inflation"]
    ],
    on="Year",
    how="inner"
)

combined_sa = combined_sa.sort_values("Year").reset_index(drop=True)

print("Final combined dataset:")
display(combined_sa)
```

### 10.2 Validating the Integrated Dataset

The final integrated dataset is validated to ensure that all years are unique, no analytical values are missing, and the variables have appropriate data types. This confirms that the dataset is ready for statistical analysis.

```python
print("FINAL DATASET VALIDATION")
print("=" * 40)

print("Number of observations:", len(combined_sa))
print("Number of duplicate rows:", combined_sa.duplicated().sum())

print("\nMissing values:")
print(combined_sa.isna().sum())

print("\nDuplicate years:", combined_sa["Year"].duplicated().sum())

print("\nData types:")
print(combined_sa.dtypes)

print("\nYear range:")
print(
    combined_sa["Year"].min(),
    "to",
    combined_sa["Year"].max()
)
```

### 10.3 Final Clean Dataset

The resulting dataset contains South African poverty and inflation observations for the years in which both indicators are available. The data is ordered chronologically and is ready for descriptive statistical analysis.

```python
display(combined_sa)
```

## Step 11: Descriptive Statistical Analysis

### 11.1 Summary Statistics

```python
descriptive_stats = combined_sa[
    ["Poverty", "Inflation"]
].describe().T

descriptive_stats
```

### 11.2 Measuring Change Over the Analysis Period

The percentage-point change between the earliest and latest common observations is calculated for both indicators to provide an initial interpretation of the long-term movement in the South African data.

```python
first_year = combined_sa.iloc[0]
last_year = combined_sa.iloc[-1]

poverty_change = last_year["Poverty"] - first_year["Poverty"]
inflation_change = last_year["Inflation"] - first_year["Inflation"]

print(f"Poverty change ({int(first_year['Year'])} to {int(last_year['Year'])}): "
      f"{poverty_change:.2f} percentage points")

print(f"Inflation change ({int(first_year['Year'])} to {int(last_year['Year'])}): "
      f"{inflation_change:.2f} percentage points")
```

### 11.3 Identifying Extreme Values

The highest and lowest observed values for poverty and inflation are identified to provide meaningful interpretation of the descriptive statistics.

```python
poverty_max = combined_sa.loc[combined_sa["Poverty"].idxmax()]
poverty_min = combined_sa.loc[combined_sa["Poverty"].idxmin()]

inflation_max = combined_sa.loc[combined_sa["Inflation"].idxmax()]
inflation_min = combined_sa.loc[combined_sa["Inflation"].idxmin()]

print("POVERTY")
print(
    f"Highest: {poverty_max['Poverty']:.1f}% "
    f"in {int(poverty_max['Year'])}"
)
print(
    f"Lowest: {poverty_min['Poverty']:.1f}% "
    f"in {int(poverty_min['Year'])}"
)

print("\nINFLATION")
print(
    f"Highest: {inflation_max['Inflation']:.3f}% "
    f"in {int(inflation_max['Year'])}"
)
print(
    f"Lowest: {inflation_min['Inflation']:.3f}% "
    f"in {int(inflation_min['Year'])}"
)
```

### 11.4 Measuring Variability

The range and interquartile range (IQR) are calculated to assess the spread of poverty and inflation observations. The IQR represents the middle 50% of the observations and is less affected by extreme values.

```python
for variable in ["Poverty", "Inflation"]:
    minimum = combined_sa[variable].min()
    maximum = combined_sa[variable].max()
    q1 = combined_sa[variable].quantile(0.25)
    q3 = combined_sa[variable].quantile(0.75)
    
    data_range = maximum - minimum
    iqr = q3 - q1

    print(variable)
    print(f"Range: {data_range:.3f}")
    print(f"IQR: {iqr:.3f}")
    print()
```

### 11.5 Consolidated Descriptive Statistics

A consolidated summary table is produced to present the main descriptive statistics for both indicators in a clear and comparable format.

```python
summary_table = pd.DataFrame({
    "Indicator": ["Poverty", "Inflation"],
    "Mean": [
        combined_sa["Poverty"].mean(),
        combined_sa["Inflation"].mean()
    ],
    "Median": [
        combined_sa["Poverty"].median(),
        combined_sa["Inflation"].median()
    ],
    "Std Dev": [
        combined_sa["Poverty"].std(),
        combined_sa["Inflation"].std()
    ],
    "Minimum": [
        combined_sa["Poverty"].min(),
        combined_sa["Inflation"].min()
    ],
    "Maximum": [
        combined_sa["Poverty"].max(),
        combined_sa["Inflation"].max()
    ]
})

summary_table.round(3)
```

### 11.6 Initial Statistical Insights

The descriptive statistics are interpreted to identify the central tendency, variability, and overall movement of poverty and inflation across the common South African observations.

```python
print("KEY FINDINGS")
print("=" * 63)

print(
    f"Average poverty: "
    f"{combined_sa['Poverty'].mean():.2f}%"
)

print(
    f"Median poverty: "
    f"{combined_sa['Poverty'].median():.2f}%"
)

print(
    f"Average inflation: "
    f"{combined_sa['Inflation'].mean():.2f}%"
)

print(
    f"Median inflation: "
    f"{combined_sa['Inflation'].median():.2f}%"
)

print(
    f"Poverty decreased by "
    f"{abs(poverty_change):.2f} percentage points "
    f"from 1993 to 2022."
)

print(
    f"Inflation decreased by "
    f"{abs(inflation_change):.2f} percentage points "
    f"from 1993 to 2022."
)
```
