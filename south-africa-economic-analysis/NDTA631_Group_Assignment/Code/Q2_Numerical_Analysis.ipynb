```python
import pandas as pd
import numpy as np

print("Pandas and NumPy are working!")
```

```python
inflation_df = pd.read_csv("inflation.csv")

print("Inflation dataset loaded successfully!")
```

```python
poverty_df = pd.read_csv("poverty.csv")

print("Poverty dataset loaded successfully!")
```

```python
print("Inflation data:")
display(inflation_df.head())

print("Poverty data:")
display(poverty_df.head())
```

```python
print("Inflation columns:")
print(inflation_df.columns.tolist())

print("\nPoverty columns:")
print(poverty_df.columns.tolist())
```

```python
print("Inflation dataset:")
display(inflation_df.head())

print("\nPoverty dataset:")
display(poverty_df.head())
```

```python
south_africa_inflation = inflation_df[
    inflation_df["REF_AREA"] == "ZAF"
]

display(south_africa_inflation)
```

```python
south_africa_poverty = poverty_df[
    poverty_df["REF_AREA"] == "ZAF"
]

display(south_africa_poverty)
```

```python
print("Inflation columns:")
print(south_africa_inflation.columns.tolist())
```

```python
print("Poverty columns:")
print(south_africa_poverty.columns.tolist())
```

```python
print(
    south_africa_inflation[
        ["REF_AREA_LABEL", "INDICATOR_LABEL", "OBS_STATUS", "OBS_STATUS_LABEL"]
    ]
)
```

# SELECT THE ACTUAL INFLATION DATA

```python
actual_inflation = south_africa_inflation[
    south_africa_inflation["OBS_STATUS_LABEL"] == "Normal value"
]

display(actual_inflation)
```

# EXTRACT THE INFLATION VALUES

```python
years = ["1993", "2000", "2005", "2008", "2010", "2014", "2022"]
```

```python
inflation_values = actual_inflation[years].iloc[0].astype(float).to_numpy()

print("Inflation values:")
print(inflation_values)
```

# EXTRACT THE POVERTY VALUES

```python
poverty_values = south_africa_poverty[years].iloc[0].astype(float).to_numpy()

print("Poverty values:")
print(poverty_values)
```

# THE NUMPY ARRAYS

```python
years_array = np.array([1993, 2000, 2005, 2008, 2010, 2014, 2022])

inflation_array = np.array(inflation_values)

poverty_array = np.array(poverty_values)

print("Years array:")
print(years_array)

print("\nInflation array:")
print(inflation_array)

print("\nPoverty array:")
print(poverty_array)
```

# THE ARRAY SHAPE

```python
print("Years shape:", years_array.shape)
print("Inflation shape:", inflation_array.shape)
print("Poverty shape:", poverty_array.shape)
```

## CALCULATING DESCRIPTIVE STATISTICS

# Inflation Statistics

```python
print("INFLATION STATISTICS")

print("Mean:", np.mean(inflation_array))
print("Median:", np.median(inflation_array))
print("Minimum:", np.min(inflation_array))
print("Maximum:", np.max(inflation_array))
print("Standard deviation:", np.std(inflation_array))
```

# Poverty Statistics

```python
print("POVERTY STATISTICS")

print("Mean:", np.mean(poverty_array))
print("Median:", np.median(poverty_array))
print("Minimum:", np.min(poverty_array))
print("Maximum:", np.max(poverty_array))
print("Standard deviation:", np.std(poverty_array))
```

```python
print("INFLATION STATISTICS")

print("Mean:", round(np.mean(inflation_array), 2))
print("Median:", round(np.median(inflation_array), 2))
print("Minimum:", round(np.min(inflation_array), 2))
print("Maximum:", round(np.max(inflation_array), 2))
print("Standard deviation:", round(np.std(inflation_array), 2))

print("\nPOVERTY STATISTICS")

print("Mean:", round(np.mean(poverty_array), 2))
print("Median:", round(np.median(poverty_array), 2))
print("Minimum:", round(np.min(poverty_array), 2))
print("Maximum:", round(np.max(poverty_array), 2))
print("Standard deviation:", round(np.std(poverty_array), 2))
```

# THE YEAR WITH THE HIGHEST INFLATION

```python
highest_inflation_index = np.argmax(inflation_array)

print("Highest inflation year:", years_array[highest_inflation_index])
print("Highest inflation value:", inflation_array[highest_inflation_index])
```

# THE YEAR WITH THE LOWEST INFLATION

```python
lowest_inflation_index = np.argmin(inflation_array)

print("Lowest inflation year:", years_array[lowest_inflation_index])
print("Lowest inflation value:", inflation_array[lowest_inflation_index])
```

# THE YEAR WITH THE HIGHEST POVERTY

```python
highest_poverty_index = np.argmax(poverty_array)

print("Highest poverty year:", years_array[highest_poverty_index])
print("Highest poverty value:", poverty_array[highest_poverty_index])
```

# THE YEAR WITH THE LOWEST POVERTY

```python
lowest_poverty_index = np.argmin(poverty_array)

print("Lowest poverty year:", years_array[lowest_poverty_index])
print("Lowest poverty value:", poverty_array[lowest_poverty_index])
```

# ARRAY OPERATIONS

```python
inflation_change = np.diff(inflation_array)

print("Change in inflation between selected years:")
print(inflation_change)
```

```python
poverty_change = np.diff(poverty_array)

print("Change in poverty between selected years:")
print(poverty_change)
```

# RESHAPING THE NUMPY ARRAYS

```python
years_reshaped = years_array.reshape(1, 7)
inflation_reshaped = inflation_array.reshape(1, 7)
poverty_reshaped = poverty_array.reshape(1, 7)

print("Years reshaped:")
print(years_reshaped)

print("\nInflation reshaped:")
print(inflation_reshaped)

print("\nPoverty reshaped:")
print(poverty_reshaped)
```

```python
print("Years shape:", years_reshaped.shape)
print("Inflation shape:", inflation_reshaped.shape)
print("Poverty shape:", poverty_reshaped.shape)
```

```python
correlation_matrix = np.corrcoef(
    inflation_array,
    poverty_array
)

print(correlation_matrix)
```

```python
correlation = correlation_matrix[0, 1]

print("Correlation between inflation and poverty:", round(correlation, 2))
```

# A SUMMARY TABLE

```python
summary = pd.DataFrame({
    "Year": years_array,
    "Inflation (%)": inflation_array,
    "Poverty (%)": poverty_array
})

display(summary)
```

# THE HIGHEST AND LOWEST VALUES

```python
print("INFLATION")
print(
    "Highest:",
    years_array[np.argmax(inflation_array)],
    "with",
    np.max(inflation_array),
    "%"
)

print(
    "Lowest:",
    years_array[np.argmin(inflation_array)],
    "with",
    np.min(inflation_array),
    "%"
)

print()

print("POVERTY")
print(
    "Highest:",
    years_array[np.argmax(poverty_array)],
    "with",
    np.max(poverty_array),
    "%"
)

print(
    "Lowest:",
    years_array[np.argmin(poverty_array)],
    "with",
    np.min(poverty_array),
    "%"
)
```

```python
print("Inflation changes between selected years:")

for i in range(len(inflation_change)):
    print(
        years_array[i],
        "to",
        years_array[i + 1],
        ":",
        round(inflation_change[i], 2),
        "percentage points"
    )

print()

print("Poverty changes between selected years:")

for i in range(len(poverty_change)):
    print(
        years_array[i],
        "to",
        years_array[i + 1],
        ":",
        round(poverty_change[i], 2),
        "percentage points"
    )
```

## EXPLAINING MY FINDINGS 

The results show that inflation in South Africa changed over the selected years. The average inflation rate was 6.66%, with the highest rate of 11.02% in 2008 and the lowest rate of 3.37% in 2005.

The poverty rate was higher in the earlier years but generally decreased over time. The highest poverty rate was 78.5% in 2000, while the lowest was 59.6% in 2022.

The correlation between inflation and poverty was 0.03, which shows a very weak relationship between the two variables. This means that inflation and poverty did not consistently increase or decrease together in the selected years.
